## Experimentos aleatorios y eventos

Un **experimento aleatorio** es un esperimento $\varepsilon$ que reproducido en condiciones idénticas puede conducir a varios resultados, el el cual no se puede predecir el resultado. 

El espacio de todos los posibles resultados de un experiemento aleatorio es llamado **espacio de estados**, y es demominado por $\Omega$. Un posible resultado de resultasos del experimento aleatorio se denota por $\omega$, donde $\omega \in \Omega$

## Espacios de probabilidad

Un **espacio de probabilidad* es una terma $(\Omega,\mathcal{F}, \mathbb{P})$, donde $\Omega$ es un conjunto no vacio, $\mathcal{F}$ es la familia de los subconjuntos de $\Omega$ y $\mathbb{P}$ es una medida de probabilidad, la cual le asigna un número no negativo a un evento, está última es una función $\mathbb{P}:\mathcal{F} \rightarrow [0,1]$ que verifica los siguientes axiomas:

1. Si $A \in \mathcal{F}$, entonces $0\leqslant \mathbb{P}(A) \leqslant 1$.
2. La probabilidad del espacio de estados $\Omega$ es igual a 1, es decir, $\mathbb{P}(\Omega)=1$.
3. Si $(A_n)_n\geqslant1$ es una sucesión finita o infinita numerable de elementos de $\mathcal{F}$, que son mutualmente excluyentes de dos a dos (es decir, $A_i\cap A_j=\emptyset$ para $i\neq j$), entonces

$$\mathbb{P} \left( \bigcup_{n=1}^\infty A_n \right)=\sum_{n=1}^\infty \mathbb{P}(A_n)$$.

**Propiedades**
1) $\mathbb{P}(\emptyset)=0$
2) $\mathbb{P}(A^c)=1-\mathbb{P}(A)$
3) $A\subset B \Rightarrow \mathbb{P}(B/A)=\mathbb{P}(B)-\mathbb{P}(A)$
4) $\mathbb{P}(A \cup B)= \mathbb{P}(A)+ \mathbb{P}(A)-\mathbb{P}(A \cap B)$


## Modelo clásico (Laplaciano)

La definición clásica de probabilidad fue introdicidad en el año 1812 por Pierre-Simon Laplace, se basó en el concepto de **equiprobabilidad** con una propiedad objetiva de los eventos asociados a un experiemtno aleatorio.

Supongamos que $\Omega = \{\omega_1,\omega_2,...\omega_n\}$ es un conjunto finito no vacio y sea $\mathcal{F}=\mathbb{P}(\Omega)$, los elementos $\{\omega_i\}$ tienen las misma probabilidad $1/n$, y se tiene que la probabilidad de un evento $A$ es:

$\mathbb{P}(A)=\frac{Casos favorables}{Casos  totales}=\frac{|A|}{|\Omega|}$, para toda $A \subset \Omega$.




## Probabilidad condicional

Sea $A, B$ eventos tal que $\mathbb{P}(B)>0$, entonces definimos la probabilidad del evento $A$ sabiendo la ocurrencia del evento $B$ como sigue:

$$\mathbb{P}(A|B)=\frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)}$$ 



## Independencia

Decimos que dos eventos $A$ y $B$ son independientes si 
$$\mathbb{P}(A\cap B)=\mathbb{P}(A)\mathbb{P}(B)$$

# Variables aleatorias

Una **variable aleatoria** real es una función $X:\Omega \rightarrow \mathbb{R}$ tal que $X^{-1}(\mathcal{B})\in \mathcal{F}$

Las varaibles aleatorias se dividen en: 
- **discretas** toma valores en un conjunto discreto (subconjunto de $\mathbb{N}, \mathbb{Z}$), peuede ser finito o no.
- **continuas** toma valores sobre un intervalo de $\mathbb{R}$.
- **mixtas** combinación de las dos anteriores. 

**Función de densidad o función de masa de probabilidad discreta** :si $X$ es v.a. discreta, entonces 

$f_x(x)=\left\{ \begin{array}{rcl}
{{\mathbb{P}(X=x)}} & ,
& si  x \in \mathbb{R}_X \\ 0 &, e.o.c. &\\
\end{array}\right.$

**Propiedades:**
1) $0\leqslant f_x (x)\leqslant 1$, #$(f_x (x)=\mathbb{P}(x))$
2) $\sum_{x \in \mathbb{R}_x}f_x (x)=1$

## Probabilidad Condicional

Si añadimos condiciones a nuestro experimento aleatorio, la probabilidad cambia.

Definimos  la probabilidad de que ocurra un evento $A$ **sabiendo** que ocurre un evento $B$, de la siguiente manera:

$$\mathbb{P}(A \mid B) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)} = \frac{|A \cap B|}{|B|},$$
donde $\mathbb{P}(B)>0$.

Queremos calcular la probabilidad condicional de los eventos:
  * $B$: el primer lanzamiento es águila
  * $C$: dos de los resultados son águila (exactamente)

````python
# Vamos a definir la función de probabilidad condicional

def prob_cond(E,F):
    return len(E & F) / len(F)
````

# Función de distribución



 Tambien conocida como función acumulada de probabilidad. 
 $$F_x:\mathbb{R} \rightarrow [0,1]$$

$$F_x (x) = \mathbb{P}(X\leqslant x) $$

### Caso discreto 
$$F_X (x)=\sum_{u\leqslant x}\mathbb{P}(x=u)$$

 

### Caso continua

Una variable aleatoria es absolutamente continua (continua) si existe

$$F_X (x)={\displaystyle \int_{-\infty}^\infty f(u)\,du}$$

**Propiedades:**
1) Monótona no decreciente: si $x \leqslant y \Rightarrow F_x (x) \leqslant F_x(y)$
2) $\lim_{x \to -\infty} F_x(x)=0$ y $\lim_{x \to \infty} F_x(x)=1$
3) $F_x$ es continua por la derecha con límites por la izquierda 

Ahora queremos resumir la informació de la v.a. $X$ mediante un valor representativo (valor esperado, esperanza matemática)

Si $X$ es una v.a real definida sobre $(\Omega, \mathcal{F}, \mathbb{P})$ 


$$
\mathbb{E}[X] =
\begin{cases}
\displaystyle \sum_{x \in R_X} x\,\mathbb{P}(X = x), & \text{si } X \text{ es discreta}, \\[6pt]
\displaystyle \int_{\mathbb{R}} x f(x)\,dx, & \text{si } X \text{ es continua}.
\end{cases}
$$

Ahora quiero saber que tan dispersos estan mis datos, esto sera la **Varianza** 
$$Var(x)=\mathbb{E}[X-\mathbb{E}(x)]= \mathbb{E}[x^2]-(\mathbb{E}(x))^2$$

### Lema del estadístico inconciente

$
\mathbb{E}[Y] = \mathbb{E}[g(X)] =
\begin{cases}
\displaystyle \sum_{x \in R_X} g(x)\,\mathbb{P}(X = x), & \text{si } X \text{ es discreta}, \\[6pt]
\displaystyle \int_{\mathbb{R}} g(x) f(x)\,dx, & \text{si } X \text{ es continua}.

# Probabilidad Frecuentista


**Jacob Bernoulli** en 1713 formuló y demostró una versión temprana de la Ley de los Grandes Números.

Sea un espacio muestral (un conjunto finito) y podemos definir la probabilidad asociada a un evento $A$ donde tengo un experimento aleatorio que se realiza $n$ veces: 

$$f_{n}(A)=\frac{n(A)}{n}$$

donde $n(A)$ es el número de veces en que ocurre $A$.

Sea el experimento aleatorio de lanzar una moneda (no justa).

````python
import random
def volado(p): #regresa la probabilidad de obtener sol
    return "sol" if random.random() < p else "águila"
    #random.random() genera un numero entre 0 y 1 
````

````python
volado (0.4) # hay un 40% de obtener sol
````

````python
# Vamos a realizar 10 volados con probilidad p=.4
vol = [volado(0.4) for i in range(10)]
vol
````

````python
# para contar el número de soles en una lista 
vol.count("sol")
````

````python
vol.count("águila")
````

Otra manera sería por medio de un bucle:

````python
resultados = [] # lista vacía
for i in range(10):
    resultado = volado(0.4)
    resultados.append(resultado) ### pongo .append para agregar a una lista
resultados
````

Ahora, vamos a simular nuestra probabilidad frecuentista: 

Un experimento con 1,000 volados (individual) y puedo repetir 500 experimentos (500 de nosotros)

````python
# Creamos una lista vacía para almacenar probabilidades (fecuencia)
p = []

for i in range(1000):  # 1000 experimentos
    N = 1000  # lanzamientos por experimento
    resultados = []

    for j in range(N):
        resultado = volado(0.63)
        resultados.append(resultado)

    n_soles = resultados.count("sol")
    p_soles = n_soles / N

    p.append(p_soles)

# promedio final (correcto)
p_soles_prom = sum(p) / len(p)

p_soles_prom
````

## **Probabilidad geométrica**

$\mathbb{P}(A)= \frac{\text{medida de la región favorable}}{\text{medida de la región total}} $ 

### **El problema del encuentro**

Romeo y Julieta plantean encontrarse en un lugar específico entre $[0,60]$ minulos (1 hora).

    - Si Romeo llega primero, entonces piensa esperar 10 minutos
    - Si Julioeta llega primero, entonces piensa esperar 10 minutos

Se elige un número al azar ente $[0,60]$ y supongamos que el par de números se elige sobre el cuadrado de modo que la probabilidad de un evento sea el área favorable entre el área total.

Sea 

$A$: Romeo y Julieta se encuentra a una distancia de 10 minutos

me gustaría encontrar la probabilidad de $A$.

````python
import numpy as np
import matplotlib.pyplot as plt

muestra= 100000

# x: tiempo de llaegada de romeo
# y: tiempk de llegada de julieta

x=np.random.uniform(0,60, muestra)
y=np.random.uniform(0,60,muestra)
# agregar la condicion encuentro

encuentro = np.where(abs(x-y)<=10, 'r', 'b')

k=0
for elemento in encuentro:
    if elemento =='r':
        k += 1 #añada
print("La probabilidad del evento A es: {}." . format(k/muestra))

plt.figure(figsize=(6,6)) # tamaño 
plt.scatter(x,y, c=encuentro, s=1)
plt.show()
````
